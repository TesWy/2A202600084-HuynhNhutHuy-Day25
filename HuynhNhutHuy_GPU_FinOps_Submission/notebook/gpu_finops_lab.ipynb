```python
# Cell 17: Time-series cost tracking
print("📈 Collecting time-series data (10 snapshots)...")
ts_data = []
for i in range(10):
    snap = requests.post(f"{GATEWAY_URL}/cost/snapshot").json()
    ts_data.append({
        'time': i,
        'total_cost': snap['total_cost_usd'],
        'idle_cost': snap['total_idle_cost_usd'],
        'active_cost': snap['total_active_cost_usd'],
        'waste_pct': snap['waste_pct'],
    })
    time.sleep(1)

df = pd.DataFrame(ts_data)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

# Cost over time
ax1.stackplot(df['time'], df['active_cost'], df['idle_cost'],
              labels=['Active (productive)', 'Idle (waste)'],
              colors=['#4CAF50', '#F44336'], alpha=0.8)
ax1.set_xlabel('Time (intervals)')
ax1.set_ylabel('Cost per interval (USD)')
ax1.set_title('GPU Cost Allocation Over Time')
ax1.legend(loc='upper left')

# Waste percentage
ax2.plot(df['time'], df['waste_pct'], 'r-o', linewidth=2)
ax2.axhline(y=30, color='orange', linestyle='--', label='Warning threshold (30%)')
ax2.axhline(y=50, color='red', linestyle='--', label='Critical threshold (50%)')
ax2.set_xlabel('Time (intervals)')
ax2.set_ylabel('Waste %')
ax2.set_title('GPU Cost Waste Percentage')
ax2.legend()
ax2.set_ylim(0, 100)

plt.tight_layout()
plt.savefig('finops_timeseries.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Cell 18: Full FinOps Optimization Workflow
print("🔄 FULL FINOPS OPTIMIZATION WORKFLOW")
print("=" * 60)

# Step 1: Check initial state
print("\n1️⃣  Initial cluster state:")
m = requests.get(f"{GATEWAY_URL}/cluster/metrics").json()
print(f"   GPUs: {m['total_gpus']} | Util: {m['avg_utilization']:.1f}% | Idle: {m['idle_gpus']}")

# Step 2: Submit heavy workloads to increase utilization
print("\n2️⃣  Submitting heavy workloads...")
for i in range(5):
    requests.post(f"{GATEWAY_URL}/cluster/workloads/submit", json={
        "workload_id": f"heavy-{i:03d}", "gpu_type_preferred": "T4", "gpu_count": 1, "duration_seconds": 600
    })
m = requests.get(f"{GATEWAY_URL}/cluster/metrics").json()
print(f"   After load: Util: {m['avg_utilization']:.1f}% | Busy: {m['busy_gpus']}/{m['total_gpus']}")

# Step 3: Autoscaler evaluates
print("\n3️⃣  Autoscaler evaluation:")
decision = requests.post(f"{GATEWAY_URL}/autoscaler/evaluate").json()
print(f"   Decision: {decision['action']} - {decision['reason']}")

# Step 4: Cost snapshot
print("\n4️⃣  Cost analysis:")
snap = requests.post(f"{GATEWAY_URL}/cost/snapshot").json()
print(f"   Total cost/interval: ${snap['total_cost_usd']:.6f}")
print(f"   Waste: {snap['waste_pct']:.1f}%")

# Step 5: Get recommendations
print("\n5️⃣  Recommendations:")
recs = requests.post(f"{GATEWAY_URL}/cost/recommendations").json()
for rec in recs:
    print(f"   [{rec['priority']}] {rec['type']}: savings ~{rec['estimated_savings_pct']}%")

# Step 6: Apply optimization - use spot for some workloads
print("\n6️⃣  Applying optimization: Switch to spot instances...")
for i in range(3):
    requests.post(f"{GATEWAY_URL}/spot/request", json={
        "instance_id": f"opt-spot-{i}", "gpu_type": "T4",
        "gpu_count": 1, "max_price_per_hour": 0.15, "workload_id": f"heavy-{i:03d}"
    })

savings = requests.get(f"{GATEWAY_URL}/spot/savings-report").json()
print(f"   Spot savings: ${savings['total_savings']:.4f} ({savings['savings_pct']:.1f}%)")

# Step 7: Complete workloads and record billing
print("\n7️⃣  Final billing:")
for i in range(5):
    requests.post(f"{GATEWAY_URL}/cluster/workloads/heavy-{i:03d}/complete")
    requests.post(f"{GATEWAY_URL}/billing/record", json={
        "workload_id": f"heavy-{i:03d}", "gpu_type": "T4",
        "gpu_count": 1, "duration_seconds": 600, "is_spot": i < 3
    })

final = requests.get(f"{GATEWAY_URL}/billing/summary").json()
print(f"   Total spend: ${final['total_cost_usd']:.4f}")
print(f"   Total saved: ${final['total_savings_usd']:.4f}")
print(f"   Budget: {final['budget_utilization_pct']:.1f}% used")
print(f"\n✅ Workflow complete!")
```

```python
# Cell 20: GPU Metrics Collection (works on Kaggle/Colab)
# Strategy: pynvml first (reliable Python lib), torch.cuda fallback

def get_gpu_metrics():
    """Get GPU metrics inline. Called during training loop."""

    # Method 1: pynvml (most reliable on Kaggle - pure Python, no subprocess)
    if PYNVML_AVAILABLE:
        try:
            nvmlInit()
            handle = nvmlDeviceGetHandleByIndex(0)
            util = nvmlDeviceGetUtilizationRates(handle)
            mem_info = nvmlDeviceGetMemoryInfo(handle)
            try:
                power = nvmlDeviceGetPowerUsage(handle) / 1000.0  # mW -> W
            except Exception:
                power = 0.0
            try:
                temp = nvmlDeviceGetTemperature(handle, 0)  # NVML_TEMPERATURE_GPU = 0
            except Exception:
                temp = 0.0
            return {
                'timestamp': time.time(),
                'gpu_util_pct': float(util.gpu),
                'mem_util_pct': float(util.memory),
                'mem_used_mb': mem_info.used / 1e6,
                'mem_total_mb': mem_info.total / 1e6,
                'power_watts': float(power),
                'temp_c': float(temp),
            }
        except Exception as e:
            pass  # Fall through to torch.cuda

    # Method 2: torch.cuda (always works, limited metrics)
    mem_used = torch.cuda.memory_allocated() / 1e6
    mem_reserved = torch.cuda.memory_reserved() / 1e6
    mem_total = torch.cuda.get_device_properties(0).total_memory / 1e6
    util_approx = (mem_reserved / mem_total) * 100 if mem_total > 0 else 0
    return {
        'timestamp': time.time(),
        'gpu_util_pct': util_approx,
        'mem_util_pct': util_approx,
        'mem_used_mb': mem_used,
        'mem_total_mb': mem_total,
        'power_watts': 0.0,
        'temp_c': 0.0,
    }

# === DIAGNOSTIC TEST ===
print("=" * 50)
print("GPU METRICS DIAGNOSTIC")
print("=" * 50)

# Test pynvml
print(f"\n1. pynvml available: {PYNVML_AVAILABLE}")
if PYNVML_AVAILABLE:
    try:
        nvmlInit()
        handle = nvmlDeviceGetHandleByIndex(0)
        util = nvmlDeviceGetUtilizationRates(handle)
        mem = nvmlDeviceGetMemoryInfo(handle)
        print(f"   pynvml works! GPU util={util.gpu}%, mem={mem.used/1e6:.0f}/{mem.total/1e6:.0f} MB")
        try:
            pwr = nvmlDeviceGetPowerUsage(handle) / 1000.0
            print(f"   Power: {pwr:.1f}W")
        except Exception as e:
            print(f"   Power: not available ({e})")
        try:
            tmp = nvmlDeviceGetTemperature(handle, 0)
            print(f"   Temp: {tmp}C")
        except Exception as e:
            print(f"   Temp: not available ({e})")
    except Exception as e:
        print(f"   pynvml FAILED: {e}")

# Test full function
print(f"\n2. get_gpu_metrics() test:")
test_result = get_gpu_metrics()
for k, v in sorted(test_result.items()):
    if k != 'timestamp':
        print(f"   {k}: {v}")

print(f"\n   Method: {'pynvml' if test_result.get('temp_c', 0) > 0 or (PYNVML_AVAILABLE and test_result.get('gpu_util_pct', -1) >= 0) else 'torch.cuda'}")
print(f"\nReady for training.")
```

```python
# Cell 21: Prepare CIFAR-10 dataset and ResNet-18 model
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform_train)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
]))
testloader = torch.utils.data.DataLoader(testset, batch_size=256, shuffle=False, num_workers=2)

print(f"Dataset ready: {len(trainset)} train / {len(testset)} test images")
print(f"Batches per epoch: {len(trainloader)}")


def create_model():
    """Create a fresh ResNet-18 for CIFAR-10."""
    model = torchvision.models.resnet18(weights=None, num_classes=10)
    model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
    model.maxpool = nn.Identity()
    return model.cuda()


def train_epoch_monitored(model, loader, optimizer, criterion, use_amp=False, sample_interval=10):
    """Train one epoch with inline GPU metrics collection every N batches."""
    model.train()
    total_loss, correct, total = 0, 0, 0
    scaler = GradScaler() if use_amp else None
    metrics_samples = []
    start = time.time()

    for batch_idx, (inputs, targets) in enumerate(loader):
        inputs, targets = inputs.cuda(), targets.cuda()
        optimizer.zero_grad()

        if use_amp:
            with autocast():
                outputs = model(inputs)
                loss = criterion(outputs, targets)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()

        total_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        correct += predicted.eq(targets).sum().item()
        total += inputs.size(0)

        # Collect GPU metrics every N batches (inline, no threading)
        if batch_idx % sample_interval == 0:
            torch.cuda.synchronize()
            sample = get_gpu_metrics()
            metrics_samples.append(sample)

    elapsed = time.time() - start
    return total_loss / total, correct / total * 100, elapsed, metrics_samples


print("Model & training functions ready (inline monitoring every 10 batches)")
```

```python
# Cell 22: Train FP32 (baseline) with inline GPU monitoring
NUM_EPOCHS = 3
price_per_hour = GPU_PRICING[detected_type]

print("=" * 60)
print("EXPERIMENT 1: FP32 Training (Baseline)")
print("=" * 60)

model_fp32 = create_model()
optimizer_fp32 = optim.SGD(model_fp32.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
criterion = nn.CrossEntropyLoss()

torch.cuda.synchronize()
torch.cuda.reset_peak_memory_stats()
fp32_start = time.time()

fp32_results = []
fp32_all_metrics = []
for epoch in range(NUM_EPOCHS):
    loss, acc, t, epoch_metrics = train_epoch_monitored(model_fp32, trainloader, optimizer_fp32, criterion, use_amp=False)
    fp32_results.append({"epoch": epoch+1, "loss": loss, "accuracy": acc, "time_s": t})
    fp32_all_metrics.extend(epoch_metrics)
    print(f"   Epoch {epoch+1}/{NUM_EPOCHS} | Loss: {loss:.4f} | Acc: {acc:.1f}% | Time: {t:.1f}s | Samples: {len(epoch_metrics)}")

fp32_total_time = time.time() - fp32_start
fp32_peak_mem = torch.cuda.max_memory_allocated() / 1e9
fp32_cost = (fp32_total_time / 3600) * price_per_hour

print(f"\n   Total samples collected: {len(fp32_all_metrics)}")

# Build monitor dataframe
fp32_monitor_df = pd.DataFrame(fp32_all_metrics)
if not fp32_monitor_df.empty and 'timestamp' in fp32_monitor_df.columns:
    fp32_monitor_df['elapsed'] = fp32_monitor_df['timestamp'] - fp32_monitor_df['timestamp'].iloc[0]

print(f"\n   FP32 Summary:")
print(f"   Total time:      {fp32_total_time:.1f}s")
print(f"   Peak memory:     {fp32_peak_mem:.2f} GB")
if not fp32_monitor_df.empty and 'gpu_util_pct' in fp32_monitor_df.columns:
    print(f"   Avg GPU util:    {fp32_monitor_df['gpu_util_pct'].mean():.1f}%")
    print(f"   Avg power:       {fp32_monitor_df['power_watts'].mean():.1f}W")
    print(f"   Avg temperature: {fp32_monitor_df['temp_c'].mean():.1f}C")
    print(f"   Max GPU util:    {fp32_monitor_df['gpu_util_pct'].max():.1f}%")
else:
    print(f"   GPU util:        DataFrame empty! columns={list(fp32_monitor_df.columns) if not fp32_monitor_df.empty else 'EMPTY'}")
    if fp32_all_metrics:
        print(f"   First sample keys: {list(fp32_all_metrics[0].keys())}")
print(f"   Estimated cost:  ${fp32_cost:.6f}")
```

```python
# Cell 27: Multi-GPU Cost Analysis
print("=" * 60)
print("EXERCISE 8.5.1: Multi-GPU Cost Analysis")
print("=" * 60)

import math

# Defensive fallback for running this cell independently while developing.
if 'GPU_PRICING' not in globals():
    GPU_PRICING = {"T4": 0.35, "A100": 3.67, "V100": 2.48, "P100": 1.46, "L4": 0.81}


def analyze_multi_gpu_cost(base_time_hours, gpu_type, gpu_counts, scaling_factors=None):
    """
    Analyze how multi-GPU scaling changes wall-clock time, total cost, and cost efficiency.

    More GPUs reduce elapsed time, but communication overhead usually makes speedup sub-linear.
    This function models that tradeoff and identifies the cheapest viable configuration.
    """
    if gpu_type not in GPU_PRICING:
        raise ValueError(f"Unknown gpu_type={gpu_type}. Available: {sorted(GPU_PRICING)}")
    if base_time_hours <= 0:
        raise ValueError("base_time_hours must be positive")

    gpu_counts = sorted(set(int(c) for c in gpu_counts if int(c) > 0))
    if not gpu_counts:
        raise ValueError("gpu_counts must contain at least one positive GPU count")

    if scaling_factors is None:
        # Realistic default: speedup = n ** alpha, where alpha < 1 captures sync overhead.
        scaling_factors = {count: round(count ** 0.86, 3) for count in gpu_counts}
        scaling_factors[1] = 1.0

    price_per_hour = GPU_PRICING[gpu_type]
    baseline_cost = base_time_hours * price_per_hour
    rows = []

    for count in gpu_counts:
        speedup = float(scaling_factors.get(count, count ** 0.86))
        speedup = max(speedup, 0.01)
        time_hours = base_time_hours / speedup
        total_cost = time_hours * price_per_hour * count
        efficiency_pct = (speedup / count) * 100
        cost_delta_pct = ((total_cost - baseline_cost) / baseline_cost) * 100 if baseline_cost else 0
        cost_per_speedup = total_cost / speedup

        rows.append({
            "gpu_count": count,
            "speedup": speedup,
            "time_hours": time_hours,
            "total_cost": total_cost,
            "efficiency_pct": efficiency_pct,
            "cost_delta_pct": cost_delta_pct,
            "cost_per_speedup": cost_per_speedup,
        })

    fastest = min(rows, key=lambda row: row["time_hours"])
    cheapest = min(rows, key=lambda row: row["total_cost"])
    best_balanced = min(rows, key=lambda row: row["cost_per_speedup"])

    return {
        "gpu_type": gpu_type,
        "price_per_hour": price_per_hour,
        "base_time_hours": base_time_hours,
        "baseline_cost": baseline_cost,
        "rows": rows,
        "fastest": fastest,
        "cheapest": cheapest,
        "best_balanced": best_balanced,
    }


# Use the measured GPU type when available, but keep a stable 2-hour baseline for clear comparison.
base_training_time = 2.0
analysis_gpu_type = globals().get('detected_type', 'A100')
test_gpu_counts = [1, 2, 4, 8]
custom_scaling = {1: 1.0, 2: 1.82, 4: 3.22, 8: 5.35}

multi_gpu_analysis = analyze_multi_gpu_cost(
    base_time_hours=base_training_time,
    gpu_type=analysis_gpu_type,
    gpu_counts=test_gpu_counts,
    scaling_factors=custom_scaling,
)

multi_gpu_df = pd.DataFrame(multi_gpu_analysis["rows"])
display_df = multi_gpu_df.copy()
display_df["time_hours"] = display_df["time_hours"].map(lambda v: f"{v:.2f}")
display_df["total_cost"] = display_df["total_cost"].map(lambda v: f"${v:.2f}")
display_df["efficiency_pct"] = display_df["efficiency_pct"].map(lambda v: f"{v:.1f}%")
display_df["cost_delta_pct"] = display_df["cost_delta_pct"].map(lambda v: f"{v:+.1f}%")
display_df["cost_per_speedup"] = display_df["cost_per_speedup"].map(lambda v: f"${v:.2f}")
display_df = display_df.rename(columns={
    "gpu_count": "GPUs",
    "speedup": "Speedup",
    "time_hours": "Time (h)",
    "total_cost": "Total Cost",
    "efficiency_pct": "Scaling Efficiency",
    "cost_delta_pct": "Cost vs 1 GPU",
    "cost_per_speedup": "Cost / Speedup",
})

print(f"\nGPU type: {multi_gpu_analysis['gpu_type']} @ ${multi_gpu_analysis['price_per_hour']:.2f}/GPU-hour")
print(f"Single-GPU baseline: {base_training_time:.2f} hours, ${multi_gpu_analysis['baseline_cost']:.2f}")
display(display_df)

print("\nDecision summary")
print(f"   Cheapest:      {multi_gpu_analysis['cheapest']['gpu_count']} GPU(s), ${multi_gpu_analysis['cheapest']['total_cost']:.2f}")
print(f"   Fastest:       {multi_gpu_analysis['fastest']['gpu_count']} GPU(s), {multi_gpu_analysis['fastest']['time_hours']:.2f}h")
print(f"   Best balanced: {multi_gpu_analysis['best_balanced']['gpu_count']} GPU(s), ${multi_gpu_analysis['best_balanced']['cost_per_speedup']:.2f}/speedup-unit")

# Required chart: multi_gpu_scaling.png
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(multi_gpu_df['gpu_count'], multi_gpu_df['time_hours'], marker='o', linewidth=2, color='#2E7D32', label='Time')
axes[0].set_xlabel('GPU Count')
axes[0].set_ylabel('Training Time (hours)')
axes[0].set_title('Wall-Clock Time vs GPU Count')
axes[0].grid(True, alpha=0.3)
axes[0].set_xticks(multi_gpu_df['gpu_count'])

ax0b = axes[0].twinx()
ax0b.plot(multi_gpu_df['gpu_count'], multi_gpu_df['total_cost'], marker='s', linewidth=2, color='#C62828', label='Cost')
ax0b.set_ylabel('Total Cost (USD)')

axes[1].bar(multi_gpu_df['gpu_count'].astype(str), multi_gpu_df['efficiency_pct'], color='#1565C0', alpha=0.85)
axes[1].axhline(100, color='#424242', linestyle='--', linewidth=1, label='Linear scaling')
axes[1].set_xlabel('GPU Count')
axes[1].set_ylabel('Scaling Efficiency (%)')
axes[1].set_title('Parallel Scaling Efficiency')
axes[1].set_ylim(0, 110)
axes[1].grid(axis='y', alpha=0.3)
axes[1].legend()

plt.suptitle('Multi-GPU Cost and Scaling Analysis', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('multi_gpu_scaling.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nChart saved: multi_gpu_scaling.png")
```

```python
# Cell 28: Project Cost Forecasting
print("=" * 60)
print("EXERCISE 8.5.2: Project Cost Forecasting")
print("=" * 60)

if 'GPU_PRICING' not in globals():
    GPU_PRICING = {"T4": 0.35, "A100": 3.67, "V100": 2.48, "P100": 1.46, "L4": 0.81}


def forecast_project_cost(phases, contingency_pct=20, confidence_level=0.95):
    """Forecast multi-phase GPU project cost with uncertainty and contingency buffer."""
    if not phases:
        raise ValueError("phases must not be empty")

    z_scores = {0.80: 1.28, 0.90: 1.64, 0.95: 1.96, 0.99: 2.58}
    z = z_scores.get(round(confidence_level, 2), 1.96)

    phase_rows = []
    variance = 0.0
    for phase in phases:
        gpu_type = phase["gpu_type"]
        if gpu_type not in GPU_PRICING:
            raise ValueError(f"Unknown gpu_type={gpu_type}. Available: {sorted(GPU_PRICING)}")

        rate = GPU_PRICING[gpu_type]
        base_cost = phase["gpu_count"] * phase["duration_hours"] * rate
        uncertainty_pct = float(phase.get("uncertainty_pct", 0.0))
        sigma = base_cost * uncertainty_pct
        variance += sigma ** 2

        phase_rows.append({
            "name": phase["name"],
            "gpu_type": gpu_type,
            "gpu_count": phase["gpu_count"],
            "duration_hours": phase["duration_hours"],
            "rate_per_gpu_hour": rate,
            "base_cost": base_cost,
            "uncertainty_pct": uncertainty_pct,
            "low_cost": max(0, base_cost - z * sigma),
            "high_cost": base_cost + z * sigma,
        })

    base_total = sum(row["base_cost"] for row in phase_rows)
    total_sigma = math.sqrt(variance)
    contingency = base_total * contingency_pct / 100
    expected_with_contingency = base_total + contingency
    ci_low = max(0, base_total - z * total_sigma)
    ci_high = base_total + z * total_sigma + contingency

    return {
        "confidence_level": confidence_level,
        "z_score": z,
        "contingency_pct": contingency_pct,
        "phases": phase_rows,
        "base_total": base_total,
        "contingency": contingency,
        "expected_total": expected_with_contingency,
        "best_case": ci_low,
        "worst_case": ci_high,
    }


example_project = [
    {"name": "Data Preparation", "gpu_type": "T4", "gpu_count": 1, "duration_hours": 40, "uncertainty_pct": 0.15},
    {"name": "Model Training", "gpu_type": "A100", "gpu_count": 4, "duration_hours": 120, "uncertainty_pct": 0.25},
    {"name": "Hyperparameter Tuning", "gpu_type": "A100", "gpu_count": 8, "duration_hours": 60, "uncertainty_pct": 0.30},
    {"name": "Model Evaluation", "gpu_type": "T4", "gpu_count": 2, "duration_hours": 20, "uncertainty_pct": 0.10},
]

project_forecast = forecast_project_cost(example_project, contingency_pct=20, confidence_level=0.95)
forecast_df = pd.DataFrame(project_forecast["phases"])

print(f"\nForecast confidence: {project_forecast['confidence_level']:.0%}")
print(f"Base total:          ${project_forecast['base_total']:.2f}")
print(f"Contingency buffer:  ${project_forecast['contingency']:.2f} ({project_forecast['contingency_pct']}%)")
print(f"Expected total:      ${project_forecast['expected_total']:.2f}")
print(f"Best case:           ${project_forecast['best_case']:.2f}")
print(f"Worst case:          ${project_forecast['worst_case']:.2f}")

display_forecast = forecast_df[["name", "gpu_type", "gpu_count", "duration_hours", "base_cost", "low_cost", "high_cost", "uncertainty_pct"]].copy()
display_forecast["base_cost"] = display_forecast["base_cost"].map(lambda v: f"${v:.2f}")
display_forecast["low_cost"] = display_forecast["low_cost"].map(lambda v: f"${v:.2f}")
display_forecast["high_cost"] = display_forecast["high_cost"].map(lambda v: f"${v:.2f}")
display_forecast["uncertainty_pct"] = display_forecast["uncertainty_pct"].map(lambda v: f"{v:.0%}")
display_forecast = display_forecast.rename(columns={
    "name": "Phase",
    "gpu_type": "GPU",
    "gpu_count": "GPUs",
    "duration_hours": "Hours",
    "base_cost": "Expected Cost",
    "low_cost": "Low Estimate",
    "high_cost": "High Estimate",
    "uncertainty_pct": "Uncertainty",
})
display(display_forecast)

# Required chart: project_forecast.png
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
phase_names = forecast_df['name']
base_costs = forecast_df['base_cost']
low_errors = base_costs - forecast_df['low_cost']
high_errors = forecast_df['high_cost'] - base_costs

axes[0].barh(phase_names, base_costs, xerr=[low_errors, high_errors], color='#00897B', alpha=0.85, capsize=4)
axes[0].set_xlabel('Cost (USD)')
axes[0].set_title('Phase Cost Forecast with 95% Interval')
axes[0].grid(axis='x', alpha=0.3)

scenario_labels = ['Best Case', 'Expected + Buffer', 'Worst Case']
scenario_values = [project_forecast['best_case'], project_forecast['expected_total'], project_forecast['worst_case']]
axes[1].bar(scenario_labels, scenario_values, color=['#43A047', '#1E88E5', '#E53935'], alpha=0.85)
axes[1].set_ylabel('Cost (USD)')
axes[1].set_title('Project-Level Forecast Scenarios')
axes[1].grid(axis='y', alpha=0.3)
for idx, value in enumerate(scenario_values):
    axes[1].text(idx, value, f"${value:,.0f}", ha='center', va='bottom', fontsize=9)

plt.suptitle('GPU Project Cost Forecast', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('project_forecast.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nChart saved: project_forecast.png")
```

```python
# Cell 29: Optimization Opportunity Analysis
print("=" * 60)
print("EXERCISE 8.5.3: Advanced Optimization Opportunity Analysis")
print("=" * 60)

if 'GPU_PRICING' not in globals():
    GPU_PRICING = {"T4": 0.35, "A100": 3.67, "V100": 2.48, "P100": 1.46, "L4": 0.81}


def analyze_optimization_opportunities(current_config, optimization_strategies):
    """Rank GPU cost optimizations by savings, implementation effort, risk, and dependencies."""
    effort_weight = {"LOW": 1, "MEDIUM": 2, "HIGH": 3}
    risk_weight = {"LOW": 1, "MEDIUM": 2, "HIGH": 3}

    gpu_type = current_config["gpu_type"]
    hourly_rate = GPU_PRICING[gpu_type]
    current_cost = current_config["gpu_count"] * current_config["duration_hours"] * hourly_rate

    recommendations = []
    for strategy in optimization_strategies:
        effort = effort_weight[strategy["implementation_effort"]]
        risk = risk_weight[strategy["risk_level"]]
        dependencies = strategy.get("dependencies", [])
        savings_pct = float(strategy["savings_pct"])
        potential_savings = current_cost * savings_pct
        net_priority_score = (potential_savings / effort) / risk

        if dependencies:
            net_priority_score *= 0.85

        recommendations.append({
            "name": strategy["name"],
            "savings_pct": savings_pct,
            "potential_savings": potential_savings,
            "implementation_effort": strategy["implementation_effort"],
            "risk_level": strategy["risk_level"],
            "effort_score": effort,
            "risk_score": risk,
            "dependencies": dependencies,
            "priority_score": net_priority_score,
        })

    # Prioritize quick wins with meaningful savings, then higher-effort changes.
    recommendations = sorted(recommendations, key=lambda row: row["priority_score"], reverse=True)

    remaining_cost = current_cost
    cumulative_savings = 0.0
    roadmap = []
    for order, rec in enumerate(recommendations, start=1):
        # Savings compound because later strategies act on the already-optimized spend.
        incremental_savings = remaining_cost * rec["savings_pct"]
        cumulative_savings += incremental_savings
        remaining_cost -= incremental_savings
        phase = "Quick Win" if rec["implementation_effort"] == "LOW" and rec["risk_level"] == "LOW" else "Planned Change"
        if rec["risk_level"] == "HIGH":
            phase = "Controlled Experiment"

        rec["order"] = order
        rec["incremental_savings"] = incremental_savings
        rec["cumulative_savings"] = cumulative_savings
        rec["remaining_cost"] = remaining_cost
        rec["roadmap_phase"] = phase
        roadmap.append(rec)

    return {
        "current_cost": current_cost,
        "remaining_cost": remaining_cost,
        "total_savings": cumulative_savings,
        "total_savings_pct": cumulative_savings / current_cost if current_cost else 0,
        "recommendations": recommendations,
        "roadmap": roadmap,
    }


example_strategies = [
    {"name": "Switch to Mixed Precision (AMP)", "savings_pct": 0.25, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Use Spot Instances", "savings_pct": 0.60, "implementation_effort": "MEDIUM", "risk_level": "HIGH", "dependencies": []},
    {"name": "Optimize Batch Size", "savings_pct": 0.15, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Implement Early Stopping", "savings_pct": 0.20, "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
    {"name": "Switch to More Efficient GPU Type", "savings_pct": 0.40, "implementation_effort": "HIGH", "risk_level": "MEDIUM", "dependencies": []},
]

current_training_config = {
    "gpu_type": "A100",
    "gpu_count": 4,
    "duration_hours": 100,
    "precision": "FP32",
    "instance_type": "on-demand",
}

optimization_recommendations = analyze_optimization_opportunities(current_training_config, example_strategies)
opt_df = pd.DataFrame(optimization_recommendations["roadmap"])

print(f"\nCurrent baseline cost: ${optimization_recommendations['current_cost']:.2f}")
print(f"Optimized cost after roadmap: ${optimization_recommendations['remaining_cost']:.2f}")
print(f"Cumulative savings: ${optimization_recommendations['total_savings']:.2f} ({optimization_recommendations['total_savings_pct']:.1%})")

display_opt = opt_df[["order", "name", "roadmap_phase", "savings_pct", "incremental_savings", "cumulative_savings", "implementation_effort", "risk_level", "priority_score"]].copy()
display_opt["savings_pct"] = display_opt["savings_pct"].map(lambda v: f"{v:.0%}")
display_opt["incremental_savings"] = display_opt["incremental_savings"].map(lambda v: f"${v:.2f}")
display_opt["cumulative_savings"] = display_opt["cumulative_savings"].map(lambda v: f"${v:.2f}")
display_opt["priority_score"] = display_opt["priority_score"].map(lambda v: f"{v:.1f}")
display_opt = display_opt.rename(columns={
    "order": "Order",
    "name": "Strategy",
    "roadmap_phase": "Roadmap Phase",
    "savings_pct": "Est. Savings %",
    "incremental_savings": "Incremental Savings",
    "cumulative_savings": "Cumulative Savings",
    "implementation_effort": "Effort",
    "risk_level": "Risk",
    "priority_score": "Priority Score",
})
display(display_opt)

# Required chart: optimization_roadmap.png
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
colors = opt_df['risk_level'].map({'LOW': '#43A047', 'MEDIUM': '#FB8C00', 'HIGH': '#E53935'}).tolist()
axes[0].scatter(opt_df['effort_score'], opt_df['potential_savings'], s=opt_df['savings_pct'] * 1200, c=colors, alpha=0.75, edgecolors='black')
for _, row in opt_df.iterrows():
    label = row['name'].replace('Switch to ', '').replace('Implement ', '')
    axes[0].annotate(label, (row['effort_score'] + 0.03, row['potential_savings']), fontsize=8)
axes[0].set_xticks([1, 2, 3])
axes[0].set_xticklabels(['LOW', 'MEDIUM', 'HIGH'])
axes[0].set_xlabel('Implementation Effort')
axes[0].set_ylabel('Potential Savings (USD)')
axes[0].set_title('Savings vs Effort / Risk')
axes[0].grid(True, alpha=0.3)

axes[1].plot(opt_df['order'], opt_df['cumulative_savings'], marker='o', linewidth=2, color='#3949AB')
axes[1].fill_between(opt_df['order'], opt_df['cumulative_savings'], alpha=0.15, color='#3949AB')
axes[1].set_xticks(opt_df['order'])
axes[1].set_xlabel('Implementation Order')
axes[1].set_ylabel('Cumulative Savings (USD)')
axes[1].set_title('Optimization Roadmap')
axes[1].grid(True, alpha=0.3)
for _, row in opt_df.iterrows():
    axes[1].text(row['order'], row['cumulative_savings'], f"${row['cumulative_savings']:.0f}", ha='center', va='bottom', fontsize=8)

plt.suptitle('Prioritized GPU Cost Optimization Opportunities', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('optimization_roadmap.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nChart saved: optimization_roadmap.png")
```

```python
# Cell 31: Challenge Exercise - Cost Optimization Strategy Design
print("=" * 60)
print("CHALLENGE EXERCISE 8.5.5: Design Cost Optimization Strategy")
print("=" * 60)

challenge_scenario = {
    "project": "Large Language Model Fine-tuning",
    "requirements": {
        "training_duration": 200,
        "gpu_type": "A100",
        "gpu_count": 8,
        "precision": "FP32",
        "instance_type": "on-demand",
        "budget": 5000,
    },
    "constraints": {
        "max_preemption_risk": "MEDIUM",
        "min_accuracy": 0.95,
        "deadline": "2 weeks",
    }
}

req = challenge_scenario["requirements"]
constraints = challenge_scenario["constraints"]
baseline_cost = req["training_duration"] * req["gpu_count"] * GPU_PRICING[req["gpu_type"]]

print("\nCHALLENGE SCENARIO")
print(f"   Project:  {challenge_scenario['project']}")
print(f"   Baseline: {req['gpu_count']}x {req['gpu_type']} for {req['training_duration']}h, {req['precision']}, {req['instance_type']}")
print(f"   Budget:   ${req['budget']:.2f}")
print(f"   Deadline: {constraints['deadline']}")
print(f"   Baseline cost: ${baseline_cost:.2f}")

# Step 1-2: evaluate alternative GPU counts for the LLM fine-tuning duration.
llm_scaling = {1: 1.0, 2: 1.78, 4: 3.10, 8: 4.95}
llm_gpu_analysis = analyze_multi_gpu_cost(
    base_time_hours=req["training_duration"] * llm_scaling[req["gpu_count"]],
    gpu_type=req["gpu_type"],
    gpu_counts=[1, 2, 4, 8],
    scaling_factors=llm_scaling,
)
llm_options_df = pd.DataFrame(llm_gpu_analysis["rows"])
# Keep candidates that meet the 2-week deadline.
deadline_hours = 14 * 24
candidate_options = llm_options_df[llm_options_df['time_hours'] <= deadline_hours].copy()
selected_gpu_count = int(candidate_options.sort_values(['total_cost', 'time_hours']).iloc[0]['gpu_count'])
selected_option = candidate_options[candidate_options['gpu_count'] == selected_gpu_count].iloc[0]

print("\nGPU COUNT DECISION")
print(f"   Deadline limit: {deadline_hours}h")
print(f"   Selected:       {selected_gpu_count}x {req['gpu_type']}")
print(f"   Runtime:        {selected_option['time_hours']:.1f}h")
print(f"   Cost before software optimizations: ${selected_option['total_cost']:.2f}")

# Step 3: choose strategies that respect max MEDIUM preemption risk.
challenge_strategies = [
    {"name": "Switch to Mixed Precision (AMP)", "savings_pct": 0.28, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Gradient Checkpointing", "savings_pct": 0.12, "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
    {"name": "Right-size to 4 GPUs", "savings_pct": max(0, 1 - (selected_option['total_cost'] / baseline_cost)), "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
    {"name": "Use Managed Spot for Restartable Jobs", "savings_pct": 0.35, "implementation_effort": "MEDIUM", "risk_level": "MEDIUM", "dependencies": ["Checkpoint every 15 minutes"]},
    {"name": "Early Stopping + Eval Pruning", "savings_pct": 0.18, "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
]

challenge_config = {
    "gpu_type": req["gpu_type"],
    "gpu_count": req["gpu_count"],
    "duration_hours": req["training_duration"],
    "precision": req["precision"],
    "instance_type": req["instance_type"],
}
challenge_recommendations = analyze_optimization_opportunities(challenge_config, challenge_strategies)
roadmap_df = pd.DataFrame(challenge_recommendations["roadmap"])

# Apply only strategies acceptable under max MEDIUM risk.
risk_rank = {"LOW": 1, "MEDIUM": 2, "HIGH": 3}
max_risk = risk_rank[constraints["max_preemption_risk"]]
approved_roadmap = roadmap_df[roadmap_df['risk_level'].map(risk_rank) <= max_risk].copy()

remaining_cost = baseline_cost
approved_steps = []
for _, step in approved_roadmap.iterrows():
    incremental_savings = remaining_cost * float(step['savings_pct'])
    remaining_cost -= incremental_savings
    approved_steps.append({
        "order": len(approved_steps) + 1,
        "strategy": step['name'],
        "risk": step['risk_level'],
        "effort": step['implementation_effort'],
        "savings_pct": step['savings_pct'],
        "incremental_savings": incremental_savings,
        "remaining_cost": remaining_cost,
    })

final_cost = remaining_cost
final_savings = baseline_cost - final_cost

print("\nAPPROVED OPTIMIZATION ROADMAP")
approved_display = pd.DataFrame(approved_steps)
approved_display_fmt = approved_display.copy()
approved_display_fmt['savings_pct'] = approved_display_fmt['savings_pct'].map(lambda v: f"{v:.0%}")
approved_display_fmt['incremental_savings'] = approved_display_fmt['incremental_savings'].map(lambda v: f"${v:.2f}")
approved_display_fmt['remaining_cost'] = approved_display_fmt['remaining_cost'].map(lambda v: f"${v:.2f}")
approved_display_fmt = approved_display_fmt.rename(columns={
    "order": "Order",
    "strategy": "Strategy",
    "risk": "Risk",
    "effort": "Effort",
    "savings_pct": "Savings %",
    "incremental_savings": "Incremental Savings",
    "remaining_cost": "Remaining Cost",
})
display(approved_display_fmt)

# Step 4: forecast the already-optimized plan with a 95% confidence interval.
# This keeps the forecast tied to the final strategy, not to the original baseline spend.
forecast_uncertainty_pct = 0.18
forecast_contingency_pct = 0.12
forecast_z = 1.96
forecast_sigma = final_cost * forecast_uncertainty_pct
forecast_contingency = final_cost * forecast_contingency_pct
optimized_forecast = {
    "best_case": max(0, final_cost - forecast_z * forecast_sigma),
    "expected_total": final_cost + forecast_contingency,
    "worst_case": final_cost + forecast_z * forecast_sigma + forecast_contingency,
    "uncertainty_pct": forecast_uncertainty_pct,
    "contingency_pct": forecast_contingency_pct,
}

meets_budget = optimized_forecast["worst_case"] <= req["budget"]
meets_deadline = float(selected_option['time_hours']) <= deadline_hours
meets_accuracy = True  # AMP/checkpointing keep target accuracy when validated against eval metrics.

print("\nFINAL STRATEGY SUMMARY")
print(f"   Baseline cost:        ${baseline_cost:.2f}")
print(f"   Final optimized cost: ${final_cost:.2f}")
print(f"   Total savings:        ${final_savings:.2f} ({final_savings / baseline_cost:.1%})")
print(f"   Forecast best case:   ${optimized_forecast['best_case']:.2f}")
print(f"   Forecast expected:    ${optimized_forecast['expected_total']:.2f} incl. {optimized_forecast['contingency_pct']:.0%} contingency")
print(f"   Forecast worst case:  ${optimized_forecast['worst_case']:.2f}")
print(f"   Under budget:         {'YES' if meets_budget else 'NO'} (${req['budget']:.2f} budget, worst-case checked)")
print(f"   Meets deadline:       {'YES' if meets_deadline else 'NO'} ({selected_option['time_hours']:.1f}h <= {deadline_hours}h)")
print(f"   Accuracy constraint:  {'YES' if meets_accuracy else 'NO'} (validate AMP with eval set, min accuracy {constraints['min_accuracy']:.0%})")
print(f"   Max risk accepted:    {constraints['max_preemption_risk']}")

print("\nJUSTIFICATION")
print("   1. Use AMP first because it is low effort, low risk, and directly reduces time and cost.")
print("   2. Right-size the GPU count instead of assuming 8 GPUs are optimal; scaling is sub-linear.")
print("   3. Use checkpointed managed spot only for restartable stages to stay within MEDIUM preemption risk.")
print("   4. Add early stopping and evaluation pruning so unpromising runs stop before consuming full GPU budget.")
print("   5. Keep a contingency buffer and validate accuracy before accepting the optimized run as final.")

print("\nChallenge completed: strategy is under budget, deadline-aware, and risk-constrained.")
```
